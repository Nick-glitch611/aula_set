```python
conjunto = {1, 2, 3, 4, 5}

# adicona apenas um elemento
conjunto.add(5)

# Usar listas no update (mais de um elemento)
conjunto.update([2, 6, 8, 4, 3])

print(conjunto)
```

```python
nomes = {'João', 'Kaio', 'Miguel', 'Cadu'}

# Ordem alfabética ou crescente
print(sorted(nomes))

#Ordem alfabética reversa ou decrescente
print(sorted(nomes, reverse=True))
```

```python
cores = {'Azul', 'Vermelho', 'Rosa', 'Verde'}

# Retira um elemento aleatório
cor = cores.pop()
print("Removido: ", cor)
print(cores)

# Remove um elemento que existe na lista
cores.remove('Rosa')

print(cores)

# Remove um elemento sem dar erro se ele não existir
cores.discard('Preto')

print(cores)

# Limpa a lista
cores.clear()

print(cores)
```

```python
conjunto_1 = {1, 4, 5}
conjunto_2 = {2, 4, 6}

# União: A | B
print(conjunto_1 | conjunto_2)

# Interseção: A & B
print(conjunto_1 & conjunto_2)

# Diferença: A - B
print(conjunto_1 - conjunto_2)

# Diferença Simétrica A ^ B
print(conjunto_1 ^ conjunto_2)
```
